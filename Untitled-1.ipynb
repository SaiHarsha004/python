```python
def emoji_converter(message):
    words = message.split(" ")
    emojis = {
        ":)": "😊",
        ":(": "😞"
    }
    output = ""
    for word in words:
        output += emojis.get(word, word) + " "
    return output

message = input(">")
print(emoji_converter(message))
```

```python
class Solution(object):
    def romanToInt(self, s):
        values = {
            'I': 1, 'V': 5, 'X': 10,
            'L': 50, 'C': 100, 'D': 500, 'M': 1000
        }

        total = 0
        prev = 0

        for char in reversed(s):
            curr = values[char]
            if curr < prev:
                total -= curr
            else:
                total += curr
            prev = curr

        return total
```

```python
prime_numbers = []
for prime_number in range(2, 101):  # Prime numbers start from 2
    is_prime = True
    for i in range(2, prime_number):
        if prime_number % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_numbers.append(prime_number)
        print(f"{prime_number} is a prime number")

print(f"\nPrime numbers found: {prime_numbers}")
```

```python
def temperature_conversion(input_unit, temp):
    if input_unit == "C":
        return (temp * 9/5) + 32  # Convert Celsius to Fahrenheit
    elif input_unit == "F":
        return (temp - 32) * 5/9  # Convert Fahrenheit to Celsius
    elif input_unit == "K":
        return temp - 273.15  # Convert Kelvin to Celsius
    else:
        return "Invalid unit"

print(temperature_conversion("F", 67))  # Output: 19.444...
```
